```python
import pandas as pd
import geopandas as gpd
from geopandas import sjoin
from shapely.geometry import Polygon
from shapely.geometry import MultiPolygon
import matplotlib.pyplot as plt
```

```python
plt.rcParams["font.family"] = 'NanumGothic'
pre_path = '../P1_Data/02.Pretreatment/'
```

```python
# 행정경계 데이터
boundary = gpd.read_file('../P1_Data/02.Pretreatment/hjd_geo2.gpkg')
boundary
```

```python
#실제값 결과
result_hjd = pd.read_csv('../P3_Result/01.CSV/A22_읍면동별_지표산출결과.csv',encoding='cp949')
#행정동코드
HJD_cd = pd.read_excel('../P1_Data/01.External/KIKcd_H.20230501.xlsx')


HJD_cd['hjd_cd'] = HJD_cd['행정동코드'].astype('str')
HJD_cd['hjd_cd'] = HJD_cd['hjd_cd'].apply(lambda x : x[0:8])
HJD_cd.rename(columns={'읍면동명': '행정기관'}, inplace=True)
HJD_cd_yang = HJD_cd[HJD_cd['시군구명'] == '양주시']
HJD_cd_yang = HJD_cd_yang[['hjd_cd','행정기관']]

#행정동별 산출값 결과에 행정동코드 매핑
result_hjd_cd = pd.merge(HJD_cd_yang,result_hjd,on='행정기관')
result_hjd_cd
```

```python
#행정동별 산출값 결과에 행정동코드 매핑
pre_boundary = pd.merge(result_hjd_cd,boundary[['hjd_cd', 'geometry']],on='hjd_cd')
pre_boundary
```
