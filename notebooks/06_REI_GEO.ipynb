```python
import pandas as pd
import geopandas as gpd
from geopandas import sjoin
from shapely.geometry import Polygon
from shapely.geometry import MultiPolygon
import matplotlib.pyplot as plt
```

```python
plt.rcParams["font.family"] = 'NanumGothic'
pre_path = '../P1_Data/02.Pretreatment/'
```

```python
# 권역경계 데이터
boundary = gpd.read_file('../P1_Data/02.Pretreatment/result_rei_boudary.gpkg')
boundary.rename(columns={'rei_nm': '권역명'}, inplace=True)
boundary
```

```python
#실제값 결과
result_rei = pd.read_csv('../P3_Result/01.CSV/A22_권역별_지표산출결과.csv',encoding='cp949')

#행정동별 산출값 결과에 행정동코드 매핑
pre_boundary = pd.merge(result_rei,boundary[['권역명', 'geometry']],on='권역명')
pre_boundary
```

```python
#인구구조
people_1r = pre_boundary[['권역명', '총인구','세대수','인구증가율(%)', '출생률(%)', '노령화지수', '인구밀도(명/km²)','기초생활수급자 비율(%)','65세이상 1인가구 비율(%)','geometry']]
#경제활력
economy_2r = pre_boundary[['권역명', '사업체수', '종사자수', '사업체 1개소당 종사자수(명/개소)','지방소득세 부과액(억원)','유동인구', '월평균_카드매출액(원)',
       '대지비율(%)', '평균공시지가(만원)', '상업지역비율(%)','사업체수 증감율(%)', '종사자수 증감율(%)', '상용근로자 비율(%)','geometry']]
#교육
education_3r = pre_boundary[['권역명','유아(0~5세)_1,000명당_보육시설수', '학령인구당_학교수(초중고)','geometry']]
#생활환경
life_4r = pre_boundary[['권역명','인구1000명당_공공문화시설수', '인구1000명당_체육시설수', '인구1000명당_공원수', '녹지율(%)',
       '상수도보급률(%)','도시가스보급률(%)', '악취배출시설 수', '미세먼지_초미세먼지','geometry']]
#보건복지
health_5r = pre_boundary[['권역명', '인구100명당_버스이용건수(건/100명)','인구1000명당_의료시설수', '인구1000명당_병상수','인구1000명당_사회복지시설수','geometry']]
#안전
safety_6r = pre_boundary[['권역명','노후주택비율', '빈집비율', '인구1000명당_교통사고건수','인구1000명당_자동차등록건수','geometry']]
#교통기반
traffic_7r = pre_boundary[['권역명', '인구100명당_버스이용건수(건/100명)', '도로집행율(%)','geometry']]
```

```python
# pre_boundary 데이터프레임을 geopandas의 GeoDataFrame으로 변환 후 저장
people_1r = gpd.GeoDataFrame(people_1r, geometry='geometry')
people_1r.to_file('../P3_Result/03.SHP/rei/people_1r.gpkg', driver='GPKG')

economy_2r = gpd.GeoDataFrame(economy_2r, geometry='geometry')
economy_2r.to_file('../P3_Result/03.SHP/rei/economy_2r.gpkg', driver='GPKG')

education_3r = gpd.GeoDataFrame(education_3r, geometry='geometry')
education_3r.to_file('../P3_Result/03.SHP/rei/education_3r.gpkg', driver='GPKG')

life_4r = gpd.GeoDataFrame(life_4r, geometry='geometry')
life_4r.to_file('../P3_Result/03.SHP/rei/life_4r.gpkg', driver='GPKG')

health_5r = gpd.GeoDataFrame(health_5r, geometry='geometry')
health_5r.to_file('../P3_Result/03.SHP/rei/health_5r.gpkg', driver='GPKG')

safety_6r = gpd.GeoDataFrame(safety_6r, geometry='geometry')
safety_6r.to_file('../P3_Result/03.SHP/rei/safety_6r.gpkg', driver='GPKG')

traffic_7r = gpd.GeoDataFrame(traffic_7r, geometry='geometry')
traffic_7r.to_file('../P3_Result/03.SHP/rei/traffic_7r.gpkg', driver='GPKG')
```
